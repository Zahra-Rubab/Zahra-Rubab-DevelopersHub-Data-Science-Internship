### Dataset Description
The **Iris Dataset** contains 150 samples of iris flowers with 4 features:
- `sepal_length` — Length of the sepal (cm)
- `sepal_width` — Width of the sepal (cm)
- `petal_length` — Length of the petal (cm)
- `petal_width` — Width of the petal (cm)
- `species` — Type of iris flower (Setosa, Versicolor, Virginica)

**Source:** Built-in Seaborn dataset — no download required

---
## Step 1: Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported successfully!')
```

---
## Step 2: Load Dataset

```python
# Load Iris dataset directly from seaborn
df = sns.load_dataset('iris')
print('Dataset loaded successfully!')
```

---
## Step 3: Dataset Understanding & Inspection

```python
# Shape of dataset
print('Shape (rows, columns):', df.shape)

# Column names
print('\nColumn Names:', df.columns.tolist())

# First 5 rows
print('\nFirst 5 Rows:')
df.head()
```

```python
# Basic statistics
print('Basic Statistics:')
df.describe()
```

```python
# Check missing values
print('Missing Values:')
print(df.isnull().sum())
print('\nNo missing values found!' if df.isnull().sum().sum() == 0 else 'Missing values exist!')
```

---
## Step 4: Data Visualization
### 4.1 Scatter Plot — Relationship between Sepal Length and Sepal Width

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title('Scatter Plot: Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.savefig('task1_scatter.png', dpi=150, bbox_inches='tight')
plt.show()
```

### 4.2 Histogram — Distribution of Features

```python
df.drop('species', axis=1).hist(figsize=(10, 6), bins=20, color='steelblue', edgecolor='black')
plt.suptitle('Histogram: Distribution of All Features', fontsize=14)
plt.tight_layout()
plt.savefig('task1_histogram.png', dpi=150, bbox_inches='tight')
plt.show()
```

### 4.3 Box Plot — Detecting Outliers and Spread

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='species', y='petal_length',
            hue='species', palette='Set2', legend=False)
plt.title('Box Plot: Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.savefig('task1_boxplot.png', dpi=150, bbox_inches='tight')
plt.show()
```
